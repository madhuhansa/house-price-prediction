### <span style=" font-weight:bold">regression imputation to fill null value of LotFrontage column</span>

> - this method fill missing values using LinearRegression with LotArea

```python
from sklearn.linear_model import LinearRegression

# Step 1: Split data into known and missing LotFrontage
known = df[df['LotFrontage'].notnull()]
missing = df[df['LotFrontage'].isnull()]

# Step 2: Prepare features (reshape is needed for sklearn)
X_train = known[['LotArea']]
y_train = known['LotFrontage']

# Step 3: Train model
model = LinearRegression()
model.fit(X_train, y_train)

# Step 4: Predict missing LotFrontage
X_missing = missing[['LotArea']]
predicted_values = model.predict(X_missing)

# Step 5: Fill in the missing LotFrontage
df.loc[df['LotFrontage'].isnull(), 'LotFrontage'] = predicted_values
```

## <span style="color:#0076cd; font-weight:bold">📌cheack outliers</span>

```python
# Finding outliers based on the Inter Quantile Range (IQR)
numeric_cols = df.select_dtypes(include=['number']).columns  # Select only numeric columns
Q1 = df[numeric_cols].quantile(0.25)
Q3 = df[numeric_cols].quantile(0.75)
IQR = Q3 - Q1

# Calculate the count of outliers
count_out = df[numeric_cols][(df[numeric_cols] < (Q1 - 1.5 * IQR)) | (df[numeric_cols] > (Q3 + 1.5 * IQR))].count()

# Create a DataFrame to display the count of outliers
df_count_out = pd.DataFrame(count_out, columns=['count_out'])
df_count_out
```
