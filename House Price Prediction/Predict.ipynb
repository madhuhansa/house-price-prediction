# <span style="color:#cd0057; font-weight:bold">⬇️Step 1: Import Essential Libraries</span>
<hr>

```python
import pandas as pd              
```

# <span style="color:#cd0057; font-weight:bold">🚚Step 2: Load Dataset</span>
<hr>

```python
test = pd.read_csv('csv_files/test_cleaned.csv')
```

# <span style="color:#cd0057; font-weight:bold">🚚Step 3: Load pkl</span>
<hr>

```python
import joblib

gbr_model = joblib.load('model_pkl/best_gbr.pkl')
```

# <span style="color:#cd0057; font-weight:bold">🤖Step 3: Step 3: Predict on test data</span>
<hr>

```python
X_test = test.drop(columns=['Id'])
```

```python
# Predict sale prices
test_predictions = gbr_model.predict(X_test)
```

# <span style="color:#cd0057; font-weight:bold">📂Step 4: save submission</span>
<hr>

```python
ids = test['Id'] 
predictions = test_predictions  

submission = pd.DataFrame({
    'Id': ids,
    'SalePrice': predictions
})
```

```python
submission.to_csv('csv_files/submission.csv', index=False)
```
