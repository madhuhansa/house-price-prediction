## <span style="color:#0076cd; font-weight:bold">📌Remove columns</span>

## <span style="color:#0076cd; font-weight:bold">📌FireplaceQu column</span>

```python
df['FireplaceQu'].value_counts()
```

```python
# Step 1: Fill missing FireplaceQu values with "NO"
df['FireplaceQu'] = df['FireplaceQu'].fillna('NO')

# Step 2: Map FireplaceQu values to numbers
quality_map = {'Ex': 5, 'Gd': 4, 'TA': 3, 'Fa': 2, 'Po': 1, 'NO': 0}
df['FireplaceQu_num'] = df['FireplaceQu'].map(quality_map)
```

```python
df.drop(columns=['FireplaceQu'], inplace=True)
```

## <span style="color:#0076cd; font-weight:bold">📌MasVnrType column</span>

```python
df['MasVnrType'] = df['MasVnrType'].fillna('NO')
```

## <span style="color:#0076cd; font-weight:bold">📌LotFrontage column</span>

### <span style=" font-weight:bold">regression imputation to fill null value of LotFrontage column</span>

```python
from sklearn.linear_model import LinearRegression

# Step 1: Split data into known and missing LotFrontage
known = df[df['LotFrontage'].notnull()]
missing = df[df['LotFrontage'].isnull()]

# Step 2: Prepare features (reshape is needed for sklearn)
X_train = known[['LotArea']]
y_train = known['LotFrontage']

# Step 3: Train model
model = LinearRegression()
model.fit(X_train, y_train)

# Step 4: Predict missing LotFrontage
X_missing = missing[['LotArea']]
predicted_values = model.predict(X_missing)

# Step 5: Fill in the missing LotFrontage
df.loc[df['LotFrontage'].isnull(), 'LotFrontage'] = predicted_values
```

## <span style="color:#0076cd; font-weight:bold">📌other columns</span>

```python
df['GarageYrBlt'] = df['GarageYrBlt'].fillna(0)
df['GarageCond'] = df['GarageCond'].fillna('NO')
df['GarageType'] = df['GarageType'].fillna('NO')
df['GarageFinish'] = df['GarageFinish'].fillna('NO')
df['GarageQual'] = df['GarageQual'].fillna('NO')
```

```python
df['BsmtFinType2'] = df['BsmtFinType2'].fillna('NO')
df['BsmtExposure'] = df['BsmtExposure'].fillna('NO')
df['BsmtFinType1'] = df['BsmtFinType1'].fillna('NO')
df['BsmtQual'] = df['BsmtQual'].fillna('NO')
df['BsmtCond'] = df['BsmtCond'].fillna('NO')
```

```python
df['MasVnrArea'] = df['MasVnrArea'].fillna(0)
```

```python
df['Electrical'] = df['Electrical'].fillna(df['Electrical'].mode()[0])
```

## <span style="color:#0076cd; font-weight:bold">📌Create new Features</span>

### <span style="color:#00cd16; font-weight:bold">📌Total Bathrooms</span>

```python
df['TotalBathrooms'] = (
    df['FullBath'] + (0.5 * df['HalfBath']) + df['BsmtFullBath'] + (0.5 * df['BsmtHalfBath'])
)
```

### <span style="color:#00cd16; font-weight:bold">📌Total porch area</span>

```python
df['TotalPorchSF'] = (
    df['OpenPorchSF'] + df['EnclosedPorch'] +
    df['3SsnPorch'] + df['ScreenPorch']
)
```

### <span style="color:#00cd16; font-weight:bold">📌TotalSF</span>

```python
df['TotalSF'] = df['1stFlrSF'] + df['2ndFlrSF'] + df['TotalBsmtSF']
```

### <span style="color:#00cd16; font-weight:bold">📌has 2ndFloor</span>

```python
df['NumFloors'] = 1 + (df['2ndFlrSF'] > 0).astype(int)
```

### <span style="color:#00cd16; font-weight:bold">📌has pool and garager</span>

```python
df['HasPool'] = df['PoolArea'].apply(lambda x: 1 if x > 0 else 0)
df['HasGarage'] = df['GarageArea'].apply(lambda x: 1 if x > 0 else 0)
```

### <span style="color:#00cd16; font-weight:bold">📌Age</span>

```python
df['Age'] = df['YrSold'] - df['YearBuilt']
```

```python
df['SinceRemodel'] = df['YrSold'] - df['YearRemodAdd']
```

```python
df['IsRemodeled'] = (df['YearBuilt'] != df['YearRemodAdd']).astype(int)
```

```python
df['IsNew'] = (df['YearBuilt'] == df['YrSold']).astype(int)
```
